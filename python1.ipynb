
给定用户列表，编写 get_active_adults(users)，返回 active=True 且 age>=18 的姓名列表。要求保持原始顺序。

```python
def get_active_users(users: list[dict]) -> list[str]:
    for user in users:
        if user["active"] and user["age"] >= 18:
            yield user["name"]

list(get_active_users(users))
```

统计词频
编写 count_words(words)，忽略大小写，返回词频字典。输入 ['AI','python','ai','Java','PYTHON']。

```python
def count_words(text: list[str])-> dict[str, int]:
    #创建空字典
    result : dict[str, int] = {};
    for word in text:
        #统一转为小写
        word = word.lower()
        result[word] = result.get(word, 0) + 1
    return result;

count_words(["Hello", "world", "hello", "Python", "python", "PYTHON"])
```

每个订单包含 user_id, amount, status。汇总每个用户已支付（PAID）的金额，结果使用 dict[int, float]。

orders = [
    {"user_id": 1, "amount": 20.0, "status": "PAID"},
    {"user_id": 2, "amount": 50.0, "status": "CANCELLED"},
    {"user_id": 1, "amount": 30.0, "status": "PAID"},
    {"user_id": 2, "amount": 10.0, "status": "PAID"},
]
# 预期：{1: 50.0, 2: 10.0}

```python
def aggregate_data(orders: list[dict]) -> dict[int, float]:
    result:dict[int, float] = {}
    for order in orders:
        if order["status"] == "PAID":
            order_id = order["user_id"]
            result[order_id] = result.get(order_id, 0) + order["amount"]
    return result

aggregate_data([
    {"user_id": 1, "amount": 20.0, "status": "PAID"},
    {"user_id": 2, "amount": 50.0, "status": "CANCELLED"},
    {"user_id": 1, "amount": 30.0, "status": "PAID"},
    {"user_id": 2, "amount": 10.0, "status": "PAID"},
])
```

实现 parse_port(value: str) -> int：输入字符串端口，允许范围 1～65535；非法数字或越界时抛 ValueError。

```python
def parse_port(value: str) -> int:
    try:
        port = int(value)
    except ValueError as e:
        raise ValueError("not port number")
    else:
        if port < 1 or port > 65535:
            raise ValueError("invalid port number")
        return port

assert parse_port("9999") == 9999 
```

编写 paginate(items, page, size)，页码从 1 开始；页码或大小非正数时抛 ValueError。对 [1,2,3,4,5]，第 2 页、每页 2 条应返回 [3,4]。

```python
def paginate(items: list[int], page: int, size: int):
    if page < 0 or size < 0:
        raise ValueError("page and size must be non-negative")
    start = (page - 1)*size
    end = start + size;
    # [a:b)取范围内的数据，切片功能
    return items[start:end]

assert paginate([1,2,3,4,5], 2, 2) == [3,4]
assert paginate([1,2,3], 9, 2) == []
```
